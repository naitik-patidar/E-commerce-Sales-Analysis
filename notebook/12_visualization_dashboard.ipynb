```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv("ecommerce_sales_analysis.csv", encoding= "latin1")
```

```python
df
```

```python
# Calculate category counts
category_counts = df['product_category'].value_counts()

# Plot pie chart
plt.figure(figsize=(6, 6))
plt.pie(
    category_counts, 
    labels=category_counts.index, 
    autopct='%1.1f%%', 
    startangle=140,
    colors=sns.color_palette('pastel')
)
plt.title('Product Category Distribution')
plt.show()
```

```python
region_count = df["region"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    region_count, 
    labels=region_count.index, 
    autopct='%1.1f%%', 
    startangle=140,
    colors=sns.color_palette('rainbow')
)
plt.title('Region wise Distribution')
plt.show()
```

# Product Cattegory Distribution By Region :

```python


plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='region', hue='product_category', palette='pastel')

plt.title('Product Category Distribution by Region', fontsize=14)
plt.xlabel('Region')
plt.ylabel('orders')
plt.legend(title='Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

# Revenue by Region : 

```python
plt.figure(figsize=(10,6))

sns.barplot(data=df, x="region", y="revenue", palette='pastel', estimator=sum, errorbar=None)

plt.title("Revenue by Region", fontsize=14)
plt.xlabel("Region")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

# Average Order Value by Region : 

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df, 
    x="region", 
    y="revenue", 
    palette="pastel", 
    estimator="mean", 
    errorbar=None
)

plt.title("Average Order Value by Region", fontsize=14)
plt.xlabel("Region")
plt.ylabel("Average Revenue")
plt.tight_layout()
plt.show()
```

# Quantity by Product Category : 

```python
plt.figure(figsize=(10,6))

sns.barplot(data= df, x= "quantity", y = "product_category", palette='pastel', estimator=sum, errorbar=None)
plt.title("Quantity by Product Category", fontsize=14)
plt.xlabel("Quantity")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

# Region × Category Heatmap 

```python
plt.figure(figsize=(10, 6))

# 1. Create a 2D pivot table (e.g., total Revenue by Region and Product Category)
heatmap_data = df.pivot_table(
    index="region", 
    columns="product_category", 
    values="revenue", 
    aggfunc="sum"
)

# 2. Plot the heatmap
sns.heatmap(
    heatmap_data,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Revenue by Region and Category", fontsize=14)
plt.xlabel("Product Category")
plt.ylabel("Region")

plt.tight_layout()
plt.show()
```
